```python
import camelot
import pandas as pd

arquivo_pdf = "arquivo.pdf"

df_final = pd.DataFrame()
total_paginas = 559
batch = 50

# Processa em blocos (batch) de 50 páginas
for inicio in range(1, total_paginas+1, batch):
    fim = min(inicio + batch-1, total_paginas)  # não ultrapassa a última página
    paginas = f"{inicio}-{fim}"
    print(f"Extraindo páginas {paginas}...")

    tabelas = camelot.read_pdf(arquivo_pdf, pages=paginas, flavor="lattice")

    if tabelas:
        bloco = pd.concat([t.df for t in tabelas], ignore_index=True)
        df_final = pd.concat([df_final, bloco], ignore_index=True)
```

```python
saida = "excel.xlsx"
df_final.to_excel(saida, index=False)
```
